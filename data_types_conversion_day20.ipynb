```python
import pandas as pd

# every column loaded as plain text, as if read from a messy CSV
df = pd.DataFrame({
    "Name": ["Asha", "Raj", "Meena", "Vikram", "Priya"],
    "Age": ["23", "25", "22", "24", "26"],
    "Salary": ["45000", "62000", "38000", "71000", "50000"],
    "Joined": ["2023-01-15", "2022-06-10", "2024-03-01", "2021-11-20", "2023-08-05"],
    "IsActive": ["True", "False", "True", "True", "False"],
    "Rating": ["4.5", "3.8", "4.9", "4.1", "3.5"],
})
df
```

## Data-type inspection report

```python
print(df.dtypes)
```

## Converting column types

```python
# 1. string to integer
df["Age"] = df["Age"].astype(int)
print(df["Age"].dtype)
```

```python
# 2. string to float
df["Rating"] = df["Rating"].astype(float)
print(df["Rating"].dtype)
```

```python
# 3. string to integer using pd.to_numeric()
df["Salary"] = pd.to_numeric(df["Salary"])
print(df["Salary"].dtype)
```

```python
# 4. string to boolean
df["IsActive"] = df["IsActive"].map({"True": True, "False": False})
print(df["IsActive"].dtype)
```

```python
# 5. string to datetime
df["Joined"] = pd.to_datetime(df["Joined"])
print(df["Joined"].dtype)
```

```python
# 6. integer back to string
df["AgeStr"] = df["Age"].astype(str)
print(df["AgeStr"].dtype)
```

## Handling conversion errors

```python
# a column with one invalid value mixed in
messy = pd.Series(["10", "20", "abc", "40"])

# astype() would raise an error here, so pd.to_numeric with errors="coerce"
# turns the bad value into NaN instead of crashing
cleaned = pd.to_numeric(messy, errors="coerce")
print(cleaned)
```

## Interview questions

**Why are correct data types important?**
Wrong types quietly break analysis. A numeric column left as text can't be summed, averaged, or compared numerically — `"45000" > "7000"` actually evaluates as a string comparison and gives the wrong answer. Dates stored as text can't be sorted chronologically or have day/month/year extracted. Correct types also reduce memory usage significantly compared to storing everything as text.

**What does astype() do?**
It converts a Series or an entire DataFrame to a specified data type, such as `.astype(int)`, `.astype(float)`, `.astype(str)`, or `.astype("datetime64[ns]")`. It raises an error immediately if any value in the column can't be converted, which makes it strict but also useful for catching bad data early.

**How can you convert a string column to numeric?**
With `.astype(int)` or `.astype(float)` when every value is guaranteed valid, or with `pd.to_numeric()` when the column might contain some invalid entries — passing `errors="coerce"` turns anything that can't be converted into `NaN` instead of crashing the whole operation, which is safer for real-world messy data.
